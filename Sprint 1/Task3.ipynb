<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Описание задачи:
Создать базовый класс ShippingOption в C#, который будет представлять
различные опции доставки. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Требования к базовому классу ShippingOption:
• Атрибуты: ID опции доставки (DeliveryOptionId), Название опции доставки
(DeliveryOptionName), Стоимость доставки (Cost).
• Методы:
o CalculateCost(): метод для расчета стоимости доставки.
o EstimateDeliveryTime(): метод для оценки времени доставки.
o GetDeliveryDetails(): метод для получения деталей опции доставки.


#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

public class ShippingOption
{
    private int _id;
    private string _name;
    private decimal _cost;

    public int DeliveryOptionId
    {
        get { return _id; }
        set { _id = value; }
    }

    public string DeliveryOptionName
    {
        get { return _name; }
        set { _name = value; }
    }

    public decimal Cost
    {
        get { return _cost; }
        set { _cost = value; }
    }

    public ShippingOption(int id, string name, decimal cost)
    {
        DeliveryOptionId = id;
        DeliveryOptionName = name;
        Cost = cost;
    }

    public virtual decimal CalculateCost()
    {
        return Cost;
    }

    public virtual string EstimateDeliveryTime()
    {
        return "1-3 дня";
    }

    public virtual string GetDeliveryDetails()
    {
        return DeliveryOptionName + ", цена: " + CalculateCost() + " руб.";
    }

    public void CompareWith(ShippingOption other)
    {
        Console.WriteLine("Сравниваем '" + DeliveryOptionName + "' и '" + other.DeliveryOptionName + "':");
        
        if (this.CalculateCost() < other.CalculateCost())
        {
            Console.WriteLine("Первый вариант дешевле!");
        }
        else
        {
            Console.WriteLine("Второй вариант дешевле или равен по цене.");
        }
    }
}

public class StandardDelivery : ShippingOption
{
    public string AverageDeliveryTime { get; set; }

    public StandardDelivery(int id, string name, decimal cost, string avgTime) : base(id, name, cost)
    {
        AverageDeliveryTime = avgTime;
    }

    public override string EstimateDeliveryTime()
    {
        return "Среднее время: " + AverageDeliveryTime;
    }
}

public class ExpressDelivery : ShippingOption
{
    public string MinDeliveryTime { get; set; }

    public ExpressDelivery(int id, string name, decimal cost, string minTime) : base(id, name, cost)
    {
        MinDeliveryTime = minTime;
    }

    public override decimal CalculateCost()
    {
        return Cost + 200;
    }
}

public class Pickup : ShippingOption
{
    public string PickupAddress { get; set; }

    public Pickup(int id, string name, decimal cost, string address) : base(id, name, cost)
    {
        PickupAddress = address;
    }

    public override string GetDeliveryDetails()
    {
        return base.GetDeliveryDetails() + " (Адрес ПВЗ: " + PickupAddress + ")";
    }
}


StandardDelivery standard = new StandardDelivery(1, "Обычная почта", 300, "5 дней");
ExpressDelivery express = new ExpressDelivery(2, "Курьер экспресс", 500, "1 день");
Pickup pickup = new Pickup(3, "Самовывоз из ПВЗ", 0, "ул. Ленина, д. 10");

Console.WriteLine(standard.GetDeliveryDetails());
Console.WriteLine("Время: " + standard.EstimateDeliveryTime());

Console.WriteLine(express.GetDeliveryDetails());
Console.WriteLine("Итоговая цена с наценкой: " + express.CalculateCost());

Console.WriteLine(pickup.GetDeliveryDetails());

Console.WriteLine();

express.CompareWith(standard);
```
