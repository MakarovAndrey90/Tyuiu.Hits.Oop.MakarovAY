### Вариант задания № 18

**Название проекта:** «Система отзывов о продуктах и услугах»

**Тема:** «Отзывы о продуктах и услугах»

**Базовый класс:** `Review`

**Производные классы:** `ServiceReview`, `ProductReview`, `ServiceTimeReview`

> В варианте третий класс назван `ServiceReview` повторно, поэтому для избежания конфликта имён он реализован как `ServiceTimeReview` — отзыв об услуге с временем начала.

<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Review в C#, который будет представлять отзывы о
продуктах или услугах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Review:
• Атрибуты: ID отзыва (ReviewId), Текст отзыва (Text), Рейтинг (Rating).
• Методы:
o DisplayReview(): метод для отображения отзыва.
o RateProduct(): метод для присвоения рейтинга продукту.
o GetReviewDetails(): метод для получения деталей отзыва.
Требования к производным классам:
1. ОтзывОбслуживания (ServiceReview): Должен содержать дополнительные
атрибуты, такие как Дата посещения (VisitDate).
Метод DisplayReview() должен быть переопределен для включения даты
посещения в отображение отзыва.
2. ОтзывТовара (ProductReview): Должен содержать дополнительные атрибуты,
такие как Идентификатор продукта (ProductId). Метод RateProduct() должен
быть переопределен для связывания рейтинга с конкретным продуктом.
3. ОтзывУслуги (ServiceReview) (если требуется третий класс): Должен
содержать дополнительные атрибуты, такие как Время начала услуги
(StartTime). Метод GetReviewDetails() должен быть переопределен для
отображения времени начала услуги вместе с другими деталями отзыва.

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

public interface IReview
{
    void DisplayReview();
    void RateProduct(int newRating);
    string GetReviewDetails();
}

public abstract class Review : IReview
{
    public int ReviewId { get; set; }
    public string Text { get; set; }
    public int Rating { get; set; }

    public string Author { get; set; }
    public DateTime CreatedAt { get; set; }
    public bool IsVerified { get; set; }
    public int HelpfulCount { get; private set; }

    protected Review(int reviewId, string text, int rating, string author)
    {
        ReviewId = reviewId;
        Text = text;
        Rating = rating;
        Author = author;
        CreatedAt = DateTime.Now;
        IsVerified = false;
        HelpfulCount = 0;
    }

    public virtual void DisplayReview()
    {
        Console.WriteLine($"[Отзыв #{ReviewId}] Автор: {Author}, рейтинг: {Rating}/5, дата: {CreatedAt:dd.MM.yyyy}");
        Console.WriteLine($"Текст: {Text}");
    }

    public virtual void RateProduct(int newRating)
    {
        Rating = newRating;
        Console.WriteLine($"Рейтинг отзыва #{ReviewId} изменён на {Rating}/5.");
    }

    public virtual string GetReviewDetails()
    {
        return $"ID: {ReviewId}, автор: {Author}, рейтинг: {Rating}/5, проверен: {(IsVerified ? "да" : "нет")}, полезно: {HelpfulCount}";
    }

    public void MarkHelpful()
    {
        HelpfulCount++;
        Console.WriteLine($"Отзыв #{ReviewId} отмечен полезным. Всего отметок: {HelpfulCount}.");
    }

    public void Verify()
    {
        IsVerified = true;
        Console.WriteLine($"Отзыв #{ReviewId} подтверждён.");
    }

    public virtual string GetSummary()
    {
        return $"#{ReviewId} — {Rating}/5 — {Text}";
    }

    public bool IsPositive()
    {
        return Rating >= 4;
    }
}

public class ServiceReview : Review
{
    public DateTime VisitDate { get; set; }
    public string ServiceName { get; set; }
    public string StaffName { get; set; }

    public ServiceReview(int reviewId, string text, int rating, string author,
        DateTime visitDate, string serviceName, string staffName)
        : base(reviewId, text, rating, author)
    {
        VisitDate = visitDate;
        ServiceName = serviceName;
        StaffName = staffName;
    }

    public override void DisplayReview()
    {
        base.DisplayReview();
        Console.WriteLine($"Услуга: {ServiceName}, сотрудник: {StaffName}, дата посещения: {VisitDate:dd.MM.yyyy}");
    }

    public override string GetReviewDetails()
    {
        return base.GetReviewDetails() + $", услуга: {ServiceName}, сотрудник: {StaffName}, дата посещения: {VisitDate:dd.MM.yyyy}";
    }

    public void RecommendService()
    {
        Console.WriteLine($"Клиент рекомендует услугу \"{ServiceName}\".");
    }

    public string GetServiceInfo()
    {
        return $"Услуга: {ServiceName}, сотрудник: {StaffName}";
    }

    public void UpdateService(string serviceName, string staffName)
    {
        ServiceName = serviceName;
        StaffName = staffName;
        Console.WriteLine($"Данные об услуге обновлены: {ServiceName}, сотрудник: {StaffName}.");
    }
}

public class ProductReview : Review, IReview
{
    public string ProductId { get; set; }
    public string ProductName { get; set; }
    public bool WouldRecommend { get; set; }

    public ProductReview(int reviewId, string text, int rating, string author,
        string productId, string productName, bool wouldRecommend)
        : base(reviewId, text, rating, author)
    {
        ProductId = productId;
        ProductName = productName;
        WouldRecommend = wouldRecommend;
    }

    public override void RateProduct(int newRating)
    {
        base.RateProduct(newRating);
        Console.WriteLine($"Рейтинг {Rating}/5 привязан к товару {ProductName} ({ProductId}).");
    }

    void IReview.RateProduct(int newRating)
    {
        Console.WriteLine("[IReview.RateProduct]");
        RateProduct(newRating);
    }

    public override string GetReviewDetails()
    {
        return base.GetReviewDetails() + $", товар: {ProductName} ({ProductId}), рекомендует: {(WouldRecommend ? "да" : "нет")}";
    }

    string IReview.GetReviewDetails()
    {
        return $"= {GetReviewDetails()}";
    }

    public void ShowProductInfo()
    {
        Console.WriteLine($"Товар: {ProductName}, артикул: {ProductId}.");
    }

    public string GetProductInfo()
    {
        return $"Товар: {ProductName}, артикул: {ProductId}";
    }

    public void UpdateProduct(string productName, string productId)
    {
        ProductName = productName;
        ProductId = productId;
        Console.WriteLine($"Данные о товаре обновлены: {ProductName} ({ProductId}).");
    }
}

public class ServiceTimeReview : Review
{
    public TimeSpan StartTime { get; set; }
    public int DurationMinutes { get; set; }
    public string Specialist { get; set; }

    public ServiceTimeReview(int reviewId, string text, int rating, string author,
        TimeSpan startTime, int durationMinutes, string specialist)
        : base(reviewId, text, rating, author)
    {
        StartTime = startTime;
        DurationMinutes = durationMinutes;
        Specialist = specialist;
    }

    public override void DisplayReview()
    {
        base.DisplayReview();
        Console.WriteLine($"Время начала услуги: {StartTime:hh\\:mm}, длительность: {DurationMinutes} мин., специалист: {Specialist}");
    }

    public override string GetReviewDetails()
    {
        return base.GetReviewDetails() + $", время начала: {StartTime:hh\\:mm}, длительность: {DurationMinutes} мин., специалист: {Specialist}";
    }

    public void ShowSchedule()
    {
        Console.WriteLine($"Услуга началась в {StartTime:hh\\:mm} и длилась {DurationMinutes} мин.");
    }

    public string GetScheduleInfo()
    {
        return $"Начало: {StartTime:hh\\:mm}, длительность: {DurationMinutes} мин.";
    }

    public void Reschedule(TimeSpan newStartTime, int newDuration)
    {
        StartTime = newStartTime;
        DurationMinutes = newDuration;
        Console.WriteLine($"Услуга перенесена на {StartTime:hh\\:mm}, длительность: {DurationMinutes} мин.");
    }
}

public class ReviewService
{
    private readonly IReview _review;

    public ReviewService(IReview review)
    {
        _review = review ?? throw new ArgumentNullException(nameof(review));
    }

    public void Publish()
    {
        Console.WriteLine("=== Публикация отзыва ===");
        _review.DisplayReview();
        Console.WriteLine(_review.GetReviewDetails());
    }

    public void UpdateRating(int newRating)
    {
        _review.RateProduct(newRating);
        Console.WriteLine($"Новый рейтинг: {_review.GetReviewDetails()}");
    }
}


ServiceReview serviceReview = new ServiceReview(
    1, "Отличный сервис, вежливый персонал.", 5, "Анна",
    new DateTime(2026, 10, 18), "Стрижка", "Мария");

ProductReview productReview = new ProductReview(
    2, "Смартфон зависает при нагрузке.", 2, "Иван",
    "PROD-9988", "Смартфон", false);

ServiceTimeReview timeReview = new ServiceTimeReview(
    3, "Всё выполнено вовремя.", 5, "Ольга",
    new TimeSpan(14, 30, 0), 45, "Елена");

List<IReview> reviews = new List<IReview> { serviceReview, productReview, timeReview };

foreach (IReview review in reviews)
{
    ReviewService reviewService = new ReviewService(review);
    reviewService.Publish();
    Console.WriteLine(new string('-', 50));
}

productReview.RateProduct(4);
IReview explicitProductReview = productReview;
explicitProductReview.RateProduct(5);
Console.WriteLine(explicitProductReview.GetReviewDetails());

serviceReview.MarkHelpful();
serviceReview.Verify();
serviceReview.RecommendService();
Console.WriteLine(serviceReview.GetServiceInfo());
Console.WriteLine($"Отзыв положительный: {(serviceReview.IsPositive() ? "да" : "нет")}");

timeReview.ShowSchedule();
Console.WriteLine(timeReview.GetScheduleInfo());
timeReview.Reschedule(new TimeSpan(15, 0, 0), 60);

productReview.ShowProductInfo();
Console.WriteLine(productReview.GetProductInfo());
productReview.UpdateProduct("Смартфон Pro", "PROD-9999");
Console.WriteLine(productReview.GetProductInfo());

Console.WriteLine(serviceReview.GetSummary());
```
